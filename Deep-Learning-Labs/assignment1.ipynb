```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-5,5,0.01)
y = np.where(x<0, 0, 1)

plt.plot(x,y)
plt.title("Binary step")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
def sigmoid(x):
  return 1/(1+np.exp(-x))

x = np.arange(-5,5,0.01)

plt.plot(x, sigmoid(x))
plt.title("Sigmoid step")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
def tanh(x):
  return (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x))

x = np.arange(-5,5,0.01)

plt.plot(x, tanh(x))
plt.title("tanh step")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
x = np.arange(-5,5,0.01)
y = np.maximum(0, np.where(x<0, 0, x))

plt.plot(x,y)
plt.title("Relu step")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
x = np.arange(-5, 5, 0.01)

# defining the activation functions:

def gelu(x):
    # using this formula: 0.5x(1 + tanh(sqrt(2/pi)(x + 0.044715x^3)))
    return 0.5 * x * (1 + np.tanh(np.sqrt(2 / np.pi) * (x + 0.044715 * np.power(x, 3))))

def softplus(x):
    return np.log(1 + np.exp(x))

def elu(x, alpha=1.0):
    return np.where(x > 0, x, alpha * (np.exp(x) - 1))

def selu(x):
    # constants for SELU
    alpha = 1.67326
    scale = 1.0507
    return scale * np.where(x > 0, x, alpha * (np.exp(x) - 1))

def leaky_relu(x, alpha=0.1):
    return np.where(x > 0, x, alpha * x)

def prelu(x, alpha=0.25):
    # parametric ReLU acts like Leaky ReLU but 'alpha' is learned
    # using a fixed alpha=0.25 here for visualization (pytorch defaults to .25)
    return np.where(x > 0, x, alpha * x)

def silu(x):
    return x / (1 + np.exp(-x))

def gaussian(x):
    return np.exp(-x**2)
```

```python
# Gaussian Error Linear Unit (GeLU)
plt.figure()
plt.plot(x, gelu(x))
plt.title("Gaussian Error Linear Unit (GeLU)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
# Softplus
plt.figure()
plt.plot(x, softplus(x))
plt.title("Softplus")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
# Exponential Linear Unit (ELU)
plt.figure()
plt.plot(x, elu(x))
plt.title("Exponential Linear Unit (ELU, alpha=1.0)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
# Scaled Exponential Linear Unit (SELU)
plt.figure()
plt.plot(x, selu(x))
plt.title("Scaled Exponential Linear Unit (SELU)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
# Leaky ReLU
plt.figure()
plt.plot(x, leaky_relu(x))
plt.title("Leaky ReLU (alpha=0.1)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
# Parametric ReLU (PReLU)
plt.figure()
plt.plot(x, prelu(x))
plt.title("Parametric ReLU (PReLU, alpha=0.25)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
# Sigmoid Linear Unit (SiLU)
plt.figure()
plt.plot(x, silu(x))
plt.title("Sigmoid Linear Unit (SiLU)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
# Gaussian
plt.figure()
plt.plot(x, gaussian(x))
plt.title("Gaussian")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```
