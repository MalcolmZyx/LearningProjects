```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-5,5,0.01)
y = np.where(x<0, 0, 1)

plt.plot(x,y)
plt.title("Binary step")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
def sigmoid(x):
  return 1/(1+np.exp(-x))

x = np.arange(-5,5,0.01)

plt.plot(x, sigmoid(x))
plt.title("Sigmoid step")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
def tanh(x):
  return (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x))

x = np.arange(-5,5,0.01)

plt.plot(x, tanh(x))
plt.title("tanh step")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

```python
x = np.arange(-5,5,0.01)
y = np.maximum(0, np.where(x<0, 0, x))

plt.plot(x,y)
plt.title("Relu step")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```
