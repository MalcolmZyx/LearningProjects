```python
from numpy import log

def squared_error(y,yhat):
  return (y-yhat)**2
```

```python
output_list = []

output_list.append(squared_error(1,1))
output_list.append(squared_error(1,0.9997))
output_list.append(squared_error(1,0.9))
output_list.append(squared_error(1,0.6))
output_list.append(squared_error(1,0.1192))
output_list.append(squared_error(0,0.1192))
# 5 more
output_list.append(squared_error(0,1))
output_list.append(squared_error(0,0.5))
output_list.append(squared_error(0,0.25))
output_list.append(squared_error(0,0.75))
output_list.append(squared_error(1,0.5))
output_list.append(squared_error(1,0.25))
output_list.append(squared_error(1,0.75))
output_list.append(squared_error(1,0.8))
output_list.append(squared_error(1,0.2))


for item in output_list:
  print(item)
```

```python
def cross_entropy(y,a):
  return -1*(y*log(a)+(1-y)*log(1-a))
```

```python
output_list = []

output_list.append(cross_entropy(1,0.9997))
output_list.append(cross_entropy(1,.9))
output_list.append(cross_entropy(1,0.6))
output_list.append(cross_entropy(1,0.1192))
output_list.append(cross_entropy(0,0.1192))
output_list.append(cross_entropy(0,1-0.1192))
# 5 more
output_list.append(cross_entropy(0,1))
output_list.append(cross_entropy(0,0.5))
output_list.append(cross_entropy(0,0.25))
output_list.append(cross_entropy(0,0.75))
output_list.append(cross_entropy(1,0.5))
output_list.append(cross_entropy(1,0.25))
output_list.append(cross_entropy(1,0.75))
output_list.append(cross_entropy(1,0.8))
output_list.append(cross_entropy(1,0.2))


for item in output_list:
  print(item)
```
