```python
import numpy as np
import pandas as pd

# here is the link to my google colab 
# https://colab.research.google.com/drive/1yMefyWgomilm8aXjZ37DIK-7u8aAu-oh?usp=sharing
from google.colab import drive
drive.mount('/content/drive')

# a. Place the following data into a file and read the data into the matrix from a file called “Data.txt”
file_name = '/content/drive/MyDrive/CS512/Data.txt'

og_matrix = np.loadtxt(file_name)
df = pd.DataFrame(og_matrix)
print("The original matrix is as follows:")
df
```

```python
# b. Select three columns, columns 3, 1, and 9, sort them in ascending order of column 3
# and make a matrix of 10 by 3 using these three columns and call it Matrix1.
matrix1 = df.iloc[:,[2,0,8]]
print("The first three random numbers are 3, 1, and 9 which makes Matrix 1 as follows:")
matrix1
```

```python
print("The sorted version of Matrix1 in ascending order of its first column (column 3 of the original matrix) is:")
matrix1_sorted = matrix1.sort_values(by=2,ascending=True)
matrix1_sorted
```

```python
# # c. Select another set of three columns, columns 5, 2, 7, sort them in descending order of
# column 5 and make a matrix of 10 by 3 using these three columns - Call it Matrix2.
matrix2 = df.iloc[:,[4,1,6]]
print("Suppose the next set of three random numbers are 5, 2, and 7. So you your output should prints:")
matrix2
```

```python
print("Sorting Matrix 2 in descending order of its first column (column 5 of the original matrix) we get:")
matrix2_sorted = matrix2.sort_values(by=4,ascending=False)
matrix2_sorted
```

```python
# d. Add the two sorted matrices to each other and put the result into another 10 by 3
# matrix call it matrix3. No need to sort Matrix 3.
Matrix1 = matrix1_sorted.to_numpy()
Matrix2 = matrix2_sorted.to_numpy()
Matrix3 = Matrix1 + Matrix2
print("Adding the contents of the Matrix1 and Matrix2, and put the result in one column Matrix called Matrix 3:")
print(Matrix3)
```

```python
# e. Add the content of each row of Matrix 3 and put it into a new 10 by 1 matrix call it Matrix 4.
Matrix4 = np.sum(Matrix3, axis=1, keepdims=True)
print("After adding the content of matrix 3 to get the matrix 4 we get:")
print(Matrix4)
```

```python
# f. Sort Matrix4 in ascending order
matrix4_sorted = np.sort(Matrix4, axis=0)
print("Sorting matrix 4 in ascending order:")
print(matrix4_sorted)
```
